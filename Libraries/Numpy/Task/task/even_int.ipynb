```python
# evenly spaced 0,1 50
ar_sp = np.linspace(0,1,50)
ar_sp
```

```python
# visulize it
import matplotlib.pyplot as p
p.figure(figsize=(10,2))
y_axis =np.zeros(50)

p.scatter(ar_sp , y_axis ,color='blue',s=30 , label='50 elements')

p.text(0,0.02 , 'first number\n (0)', ha='center',color='red',weight='bold')
p.text(
    ar_sp[1],-0.04,f' second point\n ({ar_sp[1]:.2f})',ha='center',color='green'
)

p.text(1,0.02 , '50th point\n (1)', ha='center',color='red',weight='bold')
p.title('0 to 1 evenly spaced number ',fontsize=13)
p.xlabel('values')
p.xlim(-0.05,1.05)
p.ylim(-0.06,0.06)
p.grid(True,axis='x',linestyle='--',alpha=0.6)
p.show()
```
